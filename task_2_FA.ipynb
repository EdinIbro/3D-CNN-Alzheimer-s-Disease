```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

import numpy as np
import torch.nn.functional as F
import pickle
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

import random

# ============================================================
# SEED
# ============================================================

seed = 42

random.seed(seed)
np.random.seed(seed)

torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False


# ============================================================
# CARICAMENTO DATI - TASK 2
# Aβ−Tau− vs Aβ+Tau+
# ============================================================

with open("Dataset_Edin.pkl", "rb") as f:
    data_dict = pickle.load(f)

X_fa_raw = np.array(data_dict["FA"])
X_md_raw = np.array(data_dict["MD"])
y_raw = np.array(data_dict["Labels"])

# Task 2: uso classe originale 1 e classe originale 2
mask = (y_raw == 1) | (y_raw == 2)

X_fa = X_fa_raw[mask]
X_md = X_md_raw[mask]

y = y_raw[mask].copy()

# Rimappo:
# classe originale 1 -> 0 = Aβ−Tau−
# classe originale 2 -> 1 = Aβ+Tau+
y[y == 1] = 0
y[y == 2] = 1

ANALISI_CORRENTE = "MD_TASK2"
X_input = X_md if "MD" in ANALISI_CORRENTE else X_fa

print(f"Analisi corrente: {ANALISI_CORRENTE}")
print(f"Dataset Task 2: {len(y)} campioni")
print(f"Aβ−Tau−: {np.sum(y == 0)} | Aβ+Tau+: {np.sum(y == 1)}")


# ============================================================
# DATASET CON RESIZE FISSO 3D
# ============================================================

class AlzheimerDataset(Dataset):
    def __init__(self, data, labels, target_size=(32, 128, 128)):
        self.data = data
        self.labels = labels
        self.target_size = target_size

    def __len__(self):
        return len(self.labels)

    def normalize(self, data):
        data = np.nan_to_num(data)

        lower = np.percentile(data, 1)
        upper = np.percentile(data, 99)
        data = np.clip(data, lower, upper)

        mean = data.mean()
        std = data.std()

        if std > 0:
            return (data - mean) / std

        return data

    def __getitem__(self, idx):
        vol = self.data[idx]

        vol_norm = self.normalize(vol)

        x = torch.from_numpy(vol_norm).float()

        x = x.unsqueeze(0).unsqueeze(0)

        x = F.interpolate(
            x,
            size=self.target_size,
            mode="trilinear",
            align_corners=False
        )

        x = x.squeeze(0)

        y_label = torch.tensor(self.labels[idx], dtype=torch.long)

        return x, y_label


# ============================================================
# SQUEEZE-AND-EXCITATION BLOCK 3D
# ============================================================

class SEBlock3D(nn.Module):
    def __init__(self, channels, reduction=8):
        super(SEBlock3D, self).__init__()

        self.pool = nn.AdaptiveAvgPool3d(1)

        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction),
            nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, d, h, w = x.shape

        y = self.pool(x).view(b, c)
        y = self.fc(y).view(b, c, 1, 1, 1)

        return x * y


# ============================================================
# CNN 3D CON SE BLOCK
# ============================================================

class AlzheimerCNN3D(nn.Module):
    def __init__(self, num_classes=2):
        super(AlzheimerCNN3D, self).__init__()

        self.features = nn.Sequential(
            nn.Conv3d(1, 16, kernel_size=(3, 5, 5), padding=(1, 2, 2)),
            nn.BatchNorm3d(16),
            nn.ELU(),
            SEBlock3D(16),
            nn.MaxPool3d(2),

            nn.Conv3d(16, 32, kernel_size=(3, 5, 5), padding=(1, 2, 2)),
            nn.BatchNorm3d(32),
            nn.ELU(),
            SEBlock3D(32),
            nn.MaxPool3d(2),

            nn.Conv3d(32, 64, kernel_size=(3, 5, 5), padding=(1, 2, 2)),
            nn.BatchNorm3d(64),
            nn.ELU(),
            SEBlock3D(64),
            nn.MaxPool3d(2),
        )

        self.gap = nn.AdaptiveAvgPool3d(1)

        self.classifier = nn.Sequential(
            nn.Linear(64, 32),
            nn.ELU(),
            nn.Dropout(0.3),
            nn.Linear(32, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.gap(x)
        x = torch.flatten(x, 1)
        x = self.classifier(x)
        return x


# ============================================================
# METRICHE
# ============================================================

def compute_metrics(y_true, y_pred):
    acc = accuracy_score(y_true, y_pred)
    bal_acc = balanced_accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    return acc, bal_acc, precision, recall, f1


# ============================================================
# TRAINING E VALIDATION
# ============================================================

def train_epoch(model, loader, criterion, optimizer, device):
    model.train()

    running_loss = 0.0
    all_preds = []
    all_labels = []

    for inputs, labels in loader:
        inputs = inputs.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(inputs)
        loss = criterion(outputs, labels)

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()

        _, predicted = outputs.max(1)

        running_loss += loss.item()
        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

    avg_loss = running_loss / len(loader)
    acc, bal_acc, precision, recall, f1 = compute_metrics(all_labels, all_preds)

    return avg_loss, acc, bal_acc, precision, recall, f1


def validate_epoch(model, loader, criterion, device):
    model.eval()

    running_loss = 0.0
    all_preds = []
    all_labels = []

    with torch.no_grad():
        for inputs, labels in loader:
            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model(inputs)
            loss = criterion(outputs, labels)

            _, predicted = outputs.max(1)

            running_loss += loss.item()
            all_preds.extend(predicted.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    avg_loss = running_loss / len(loader)
    acc, bal_acc, precision, recall, f1 = compute_metrics(all_labels, all_preds)

    return avg_loss, acc, bal_acc, precision, recall, f1


# ============================================================
# PLOT CURVE
# ============================================================

def plot_fold_curves(
    train_losses,
    val_losses,
    train_f1s,
    val_f1s,
    train_accs,
    val_accs,
    fold,
    modality
):
    epochs_range = range(1, len(train_losses) + 1)

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))

    axes[0].plot(epochs_range, train_losses, label="Training Loss")
    axes[0].plot(epochs_range, val_losses, label="Validation Loss")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss")
    axes[0].set_title(f"Fold {fold} - Loss")
    axes[0].legend()
    axes[0].grid(True)

    axes[1].plot(epochs_range, train_accs, label="Training Accuracy")
    axes[1].plot(epochs_range, val_accs, label="Validation Accuracy")
    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Accuracy")
    axes[1].set_title(f"Fold {fold} - Accuracy")
    axes[1].legend()
    axes[1].grid(True)

    axes[2].plot(epochs_range, train_f1s, label="Training Macro F1")
    axes[2].plot(epochs_range, val_f1s, label="Validation Macro F1")
    axes[2].set_xlabel("Epoch")
    axes[2].set_ylabel("Macro F1")
    axes[2].set_title(f"Fold {fold} - Macro F1")
    axes[2].legend()
    axes[2].grid(True)

    plt.suptitle(f"{modality} - Training vs Validation - Fold {fold}")
    plt.tight_layout()
    plt.show()


# ============================================================
# IPERPARAMETRI
# ============================================================

epochs = 50
learning_rate = 1e-4
weight_decay = 1e-4
k_folds = 5
batch_size = 8
patience = 12

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print(f"Device utilizzato: {device}")

skf = StratifiedKFold(
    n_splits=k_folds,
    shuffle=True,
    random_state=42
)

splits = list(skf.split(X_input, y))
fold_results = []


# ============================================================
# CROSS-VALIDATION
# ============================================================

for fold, (train_ids, val_ids) in enumerate(splits, start=1):
    print(f"\n{'=' * 60}")
    print(f"INIZIO FOLD {fold}/{k_folds} - {ANALISI_CORRENTE}")
    print(f"{'=' * 60}")

    train_ds = AlzheimerDataset(
        X_input[train_ids],
        y[train_ids],
        target_size=(32, 128, 128)
    )

    val_ds = AlzheimerDataset(
        X_input[val_ids],
        y[val_ids],
        target_size=(32, 128, 128)
    )

    train_loader = DataLoader(
        train_ds,
        batch_size=batch_size,
        shuffle=True
    )

    val_loader = DataLoader(
        val_ds,
        batch_size=batch_size,
        shuffle=False
    )

    model = AlzheimerCNN3D(num_classes=2).to(device)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=learning_rate,
        weight_decay=weight_decay
    )

    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=8
    )

    n_class0 = np.sum(y[train_ids] == 0)
    n_class1 = np.sum(y[train_ids] == 1)

    weights_class = torch.tensor(
        [
            len(train_ids) / (2 * n_class0),
            len(train_ids) / (2 * n_class1)
        ],
        dtype=torch.float32
    ).to(device)

    print(f"Pesi classi fold {fold}: {weights_class.cpu().numpy()}")
    weights_class = weights_class / weights_class.sum()

    criterion = nn.CrossEntropyLoss(
        weight=weights_class,
    )

    best_val_bal_acc = 0.0
    best_epoch = 0
    best_metrics = None
    epochs_no_improve = 0

    train_losses = []
    val_losses = []

    train_accs = []
    val_accs = []

    train_bal_accs = []
    val_bal_accs = []

    train_precisions = []
    val_precisions = []

    train_recalls = []
    val_recalls = []

    train_f1s = []
    val_f1s = []

    for epoch in range(epochs):
        train_loss, train_acc, train_bal_acc, train_precision, train_recall, train_f1 = train_epoch(
            model,
            train_loader,
            criterion,
            optimizer,
            device
        )

        val_loss, val_acc, val_bal_acc, val_precision, val_recall, val_f1 = validate_epoch(
            model,
            val_loader,
            criterion,
            device
        )

        scheduler.step(val_loss)

        train_losses.append(train_loss)
        val_losses.append(val_loss)

        train_accs.append(train_acc)
        val_accs.append(val_acc)

        train_bal_accs.append(train_bal_acc)
        val_bal_accs.append(val_bal_acc)

        train_precisions.append(train_precision)
        val_precisions.append(val_precision)

        train_recalls.append(train_recall)
        val_recalls.append(val_recall)

        train_f1s.append(train_f1)
        val_f1s.append(val_f1)

        if val_bal_acc > best_val_bal_acc:
            best_val_bal_acc = val_bal_acc
            best_epoch = epoch + 1
            epochs_no_improve = 0

            best_metrics = {
                "fold": fold,
                "best_epoch": best_epoch,
                "best_val_loss": val_loss,
                "best_val_acc": val_acc,
                "best_val_bal_acc": val_bal_acc,
                "best_val_precision": val_precision,
                "best_val_recall": val_recall,
                "best_val_f1": val_f1
            }

            torch.save(
                model.state_dict(),
                f"best_model_{ANALISI_CORRENTE}_fold{fold}.pth"
            )

        else:
            epochs_no_improve += 1

        if (epoch + 1) % 10 == 0:
            print(
                f"\nFold {fold} | Epoca {epoch + 1:03d}/{epochs} | "
                f"T-Loss: {train_loss:.3f} | "
                f"T-Acc: {train_acc:.3f} | "
                f"T-BalAcc: {train_bal_acc:.3f} | "
                f"T-Prec: {train_precision:.3f} | "
                f"T-Rec: {train_recall:.3f} | "
                f"T-F1: {train_f1:.3f} || "
                f"V-Loss: {val_loss:.3f} | "
                f"V-Acc: {val_acc:.3f} | "
                f"V-BalAcc: {val_bal_acc:.3f} | "
                f"V-Prec: {val_precision:.3f} | "
                f"V-Rec: {val_recall:.3f} | "
                f"V-F1: {val_f1:.3f}"
            )
        elif (epoch + 1) % 2 == 0:
            print(".", end="", flush=True)

        if epochs_no_improve >= patience:
            print(f"\nEarly stopping attivato al fold {fold}, epoca {epoch + 1}")
            break

    fold_results.append(best_metrics)

    print(f"\nFine Fold {fold}")
    print(f"Best Epoch: {best_epoch}")
    print(f"Best Validation Balanced ACC: {best_val_bal_acc:.3f}")
    print(f"Validation Acc alla best epoch: {best_metrics['best_val_acc']:.3f}")
    print(f"Validation Balanced Acc alla best epoch: {best_metrics['best_val_bal_acc']:.3f}")

    plot_fold_curves(
        train_losses,
        val_losses,
        train_f1s,
        val_f1s,
        train_accs,
        val_accs,
        fold,
        ANALISI_CORRENTE
    )


# ============================================================
# RISULTATI FINALI CROSS-VALIDATION
# ============================================================

print("\n" + "=" * 60)
print(f"RISULTATI FINALI CROSS-VALIDATION - {ANALISI_CORRENTE}")
print("=" * 60)

best_f1s = [r["best_val_f1"] for r in fold_results]
best_accs = [r["best_val_acc"] for r in fold_results]
best_bal_accs = [r["best_val_bal_acc"] for r in fold_results]
best_precs = [r["best_val_precision"] for r in fold_results]
best_recs = [r["best_val_recall"] for r in fold_results]

print(f"Best F1 per fold: {[round(x, 3) for x in best_f1s]}")
print(f"ACC alla best epoch per fold: {[round(x, 3) for x in best_accs]}")
print(f"Balanced ACC alla best epoch per fold: {[round(x, 3) for x in best_bal_accs]}")
print(f"PREC alla best epoch per fold: {[round(x, 3) for x in best_precs]}")
print(f"REC alla best epoch per fold: {[round(x, 3) for x in best_recs]}")

print(f"\nF1 medio: {np.mean(best_f1s):.3f} ± {np.std(best_f1s):.3f}")
print(f"ACC media: {np.mean(best_accs):.3f} ± {np.std(best_accs):.3f}")
print(f"Balanced ACC media: {np.mean(best_bal_accs):.3f} ± {np.std(best_bal_accs):.3f}")
print(f"PREC media: {np.mean(best_precs):.3f} ± {np.std(best_precs):.3f}")
print(f"REC media: {np.mean(best_recs):.3f} ± {np.std(best_recs):.3f}")


# ============================================================
# REPORT FINALE AGGREGATO SUI FOLD DI VALIDAZIONE
# ============================================================

def generate_final_text_report(X_data, y, device, modality, splits):
    all_preds = []
    all_labels = []

    for fold, (_, val_ids) in enumerate(splits, start=1):
        model = AlzheimerCNN3D(num_classes=2).to(device)

        model.load_state_dict(
            torch.load(
                f"best_model_{modality}_fold{fold}.pth",
                map_location=device,
                weights_only=True
            )
        )

        model.eval()

        val_ds = AlzheimerDataset(
            X_data[val_ids],
            y[val_ids],
            target_size=(32, 128, 128)
        )

        val_loader = DataLoader(
            val_ds,
            batch_size=1,
            shuffle=False
        )

        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs = inputs.to(device)

                outputs = model(inputs)
                _, predicted = outputs.max(1)

                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(labels.numpy())

    cm = confusion_matrix(all_labels, all_preds)

    print(f"\n{'=' * 20} REPORT FINALE {modality} {'=' * 20}")
    print(
        classification_report(
            all_labels,
            all_preds,
            target_names=["Aβ−Tau−", "Aβ+Tau+"],
            zero_division=0
        )
    )

    print(f"Accuracy finale aggregata: {accuracy_score(all_labels, all_preds):.3f}")
    print(f"Balanced accuracy finale aggregata: {balanced_accuracy_score(all_labels, all_preds):.3f}")
    print(f"Macro F1 finale aggregato: {f1_score(all_labels, all_preds, average='macro', zero_division=0):.3f}")

    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Aβ−Tau−", "Aβ+Tau+"],
        yticklabels=["Aβ−Tau−", "Aβ+Tau+"]
    )
    plt.xlabel("Predetto")
    plt.ylabel("Reale")
    plt.title(f"Matrice di Confusione Finale: {modality}")
    plt.show()


generate_final_text_report(
    X_input,
    y,
    device,
    ANALISI_CORRENTE,
    splits
)
```
