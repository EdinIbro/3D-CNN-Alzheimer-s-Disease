```python
info = data["INFO"]

print("Number of subjects:", len(info))

print("\nAge")
print(info["Age_Visit"].describe())

print("\nSex")
print(info["PTGENDER"].value_counts())

print("\nMMSE")
print(info["MMSE"].describe())

print("\nClinical diagnosis")
print(info["DX"].value_counts())
```

```python
import pandas as pd

info = data["INFO"].copy()
info["Label"] = data["Labels"]

groups = {
    0: "Aβ−τ−",
    2: "Aβ+τ+"
}

rows = []

for label, name in groups.items():
    df = info[info["Label"] == label]

    n = len(df)

    dx = df["DX"].value_counts()
    cn = dx.get("CN", 0)
    mci = dx.get("MCI", 0)
    ad = dx.get("Dementia", 0)

    age = f"{df['Age_Visit'].mean():.1f} ± {df['Age_Visit'].std():.1f}"

    males = (df["PTGENDER"] == "Male").sum()
    females = (df["PTGENDER"] == "Female").sum()

    rows.append([
        name,
        n,
        f"{cn}/{mci}/{ad}",
        age,
        f"{males}/{females}"
    ])

table = pd.DataFrame(
    rows,
    columns=[
        "CLASS",
        "# OF SUBJECTS",
        "CN/MCI/AD",
        "AGE",
        "SEX (M/F)"
    ]
)

print(table)
```
