```python
# ============================================================
# IMPORT
# ============================================================

import os
import random
import pickle

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

from captum.attr import Saliency

from nilearn import plotting, image, masking
import nibabel as nib


# ============================================================
# SEED
# ============================================================

seed = 42

random.seed(seed)
np.random.seed(seed)

torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False


# ============================================================
# CARICAMENTO DATI
# ============================================================

with open("Dataset_Edin.pkl", "rb") as f:
    data_dict = pickle.load(f)

X_fa_raw = np.array(data_dict["FA"])
X_md_raw = np.array(data_dict["MD"])
y_raw = np.array(data_dict["Labels"])

mask = (y_raw == 0) | (y_raw == 2)

X_fa = X_fa_raw[mask]
X_md = X_md_raw[mask]

y = y_raw[mask].copy()
y[y == 2] = 1

ANALISI_CORRENTE = "FA"   # cambia in "MD" per usare MD

X_input = X_fa if ANALISI_CORRENTE == "FA" else X_md

print(f"Analisi corrente: {ANALISI_CORRENTE}")
print(f"Dataset: {len(y)} campioni")
print(f"Sani: {np.sum(y == 0)} | Alzheimer: {np.sum(y == 1)}")
print(f"Shape input: {X_input.shape}")


# ============================================================
# DATASET
# ============================================================

class CustomDataset(Dataset):
    def __init__(self, data, labels):
        self.data = torch.tensor(data, dtype=torch.float32)
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        x = self.data[idx]

        if x.ndim == 3:
            x = x.unsqueeze(0)

        return x, self.labels[idx]


# ============================================================
# CNN 3D
# ============================================================

class AlzheimerCNN3D(nn.Module):
    def __init__(self, num_classes=2):
        super(AlzheimerCNN3D, self).__init__()

        self.features = nn.Sequential(
            nn.Conv3d(1, 16, kernel_size=3, padding=1),
            nn.BatchNorm3d(16),
            nn.ReLU(inplace=True),
            nn.MaxPool3d(2),

            nn.Conv3d(16, 16, kernel_size=3, padding=1),
            nn.BatchNorm3d(16),
            nn.ReLU(inplace=True),
            nn.MaxPool3d(2),

            nn.Conv3d(16, 32, kernel_size=3, padding=1),
            nn.BatchNorm3d(32),
            nn.ReLU(inplace=True),
            nn.MaxPool3d(2),

            nn.Conv3d(32, 32, kernel_size=3, padding=1),
            nn.BatchNorm3d(32),
            nn.ReLU(inplace=True),
            nn.MaxPool3d(2),

            nn.Conv3d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm3d(64),
            nn.ReLU(inplace=True),

            nn.AdaptiveMaxPool3d((2, 3, 2))
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64 * 2 * 3 * 2, 128),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x


# ============================================================
# METRICHE
# ============================================================

def compute_metrics(y_true, y_pred):
    acc = accuracy_score(y_true, y_pred)
    bal_acc = balanced_accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    return acc, bal_acc, precision, recall, f1


def predict_with_threshold(outputs, threshold=0.5):
    probs = torch.softmax(outputs, dim=1)
    ad_probs = probs[:, 1]
    predicted = (ad_probs >= threshold).long()
    return predicted


# ============================================================
# TRAINING
# ============================================================

def train_epoch(model, loader, criterion, optimizer, device):
    model.train()

    running_loss = 0.0
    all_preds = []
    all_labels = []

    for inputs, labels in loader:
        inputs = inputs.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(inputs)
        loss = criterion(outputs, labels)

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        _, predicted = outputs.max(1)

        running_loss += loss.item()
        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

    avg_loss = running_loss / len(loader)

    acc, bal_acc, precision, recall, f1 = compute_metrics(
        all_labels,
        all_preds
    )

    return avg_loss, acc, bal_acc, precision, recall, f1


# ============================================================
# VALIDATION CON THRESHOLD
# ============================================================

def validate_epoch(model, loader, criterion, device, threshold=0.5):
    model.eval()

    running_loss = 0.0
    all_preds = []
    all_labels = []

    with torch.no_grad():
        for inputs, labels in loader:
            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model(inputs)
            loss = criterion(outputs, labels)

            predicted = predict_with_threshold(outputs, threshold)

            running_loss += loss.item()
            all_preds.extend(predicted.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    avg_loss = running_loss / len(loader)

    acc, bal_acc, precision, recall, f1 = compute_metrics(
        all_labels,
        all_preds
    )

    return avg_loss, acc, bal_acc, precision, recall, f1


# ============================================================
# PLOT CURVE
# ============================================================

def plot_fold_curves(
    train_losses,
    val_losses,
    train_f1s,
    val_f1s,
    train_accs,
    val_accs,
    fold,
    modality
):
    epochs_range = range(1, len(train_losses) + 1)

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))

    axes[0].plot(epochs_range, train_losses, label="Training Loss")
    axes[0].plot(epochs_range, val_losses, label="Validation Loss")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss")
    axes[0].set_title(f"Fold {fold} - Loss")
    axes[0].legend()
    axes[0].grid(True)

    axes[1].plot(epochs_range, train_accs, label="Training Accuracy")
    axes[1].plot(epochs_range, val_accs, label="Validation Accuracy")
    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Accuracy")
    axes[1].set_title(f"Fold {fold} - Accuracy")
    axes[1].legend()
    axes[1].grid(True)

    axes[2].plot(epochs_range, train_f1s, label="Training Macro F1")
    axes[2].plot(epochs_range, val_f1s, label="Validation Macro F1")
    axes[2].set_xlabel("Epoch")
    axes[2].set_ylabel("Macro F1")
    axes[2].set_title(f"Fold {fold} - Macro F1")
    axes[2].legend()
    axes[2].grid(True)

    plt.suptitle(f"{modality} - Fold {fold}")
    plt.tight_layout()
    plt.show()


# ============================================================
# XAI - SALIENCY MAPS
# ============================================================

def compute_saliency_map(model, input_tensor, target_class, device):
    model.eval()

    x = input_tensor.unsqueeze(0).to(device)
    x.requires_grad_(True)

    saliency = Saliency(model)

    attribution = saliency.attribute(
        x,
        target=int(target_class)
    )

    saliency_map = attribution.abs().detach().cpu().squeeze().numpy()

    return saliency_map


def compute_mean_saliency_maps_for_fold(
    model,
    X_data,
    y,
    val_ids,
    fold,
    modality,
    device,
    threshold,
    base_dir="saliency_maps"
):
    """
    Calcola la saliency map per tutti i soggetti del validation set
    classificati correttamente.

    Poi calcola:
    - mean_saliency_healthy
    - mean_saliency_ad
    """

    print(f"\nCalcolo saliency medie - Fold {fold}")

    output_dir = os.path.join(base_dir, modality, f"fold_{fold}")
    os.makedirs(output_dir, exist_ok=True)

    val_ds = CustomDataset(
        X_data[val_ids],
        y[val_ids]
    )

    healthy_saliency_maps = []
    ad_saliency_maps = []

    n_healthy_correct = 0
    n_ad_correct = 0

    for i in range(len(val_ds)):
        x, label = val_ds[i]

        with torch.no_grad():
            output = model(x.unsqueeze(0).to(device))
            pred = predict_with_threshold(
                output,
                threshold=threshold
            )

        true_label = int(label.item())
        pred_label = int(pred.item())

        # uso solo soggetti classificati correttamente
        if true_label != pred_label:
            continue

        saliency_map = compute_saliency_map(
            model=model,
            input_tensor=x,
            target_class=pred_label,
            device=device
        )

        if true_label == 0:
            healthy_saliency_maps.append(saliency_map)
            n_healthy_correct += 1

        elif true_label == 1:
            ad_saliency_maps.append(saliency_map)
            n_ad_correct += 1

    print(f"Fold {fold} - Healthy corretti usati: {n_healthy_correct}")
    print(f"Fold {fold} - AD corretti usati: {n_ad_correct}")

    mean_saliency_healthy = None
    mean_saliency_ad = None

    if len(healthy_saliency_maps) > 0:
        mean_saliency_healthy = np.mean(
            np.stack(healthy_saliency_maps, axis=0),
            axis=0
        )

        np.save(
            os.path.join(output_dir, f"{modality}_fold_{fold}_mean_saliency_healthy.npy"),
            mean_saliency_healthy
        )

    if len(ad_saliency_maps) > 0:
        mean_saliency_ad = np.mean(
            np.stack(ad_saliency_maps, axis=0),
            axis=0
        )

        np.save(
            os.path.join(output_dir, f"{modality}_fold_{fold}_mean_saliency_AD.npy"),
            mean_saliency_ad
        )

    return mean_saliency_healthy, mean_saliency_ad


def display_brain_maps(
    image_roi: np.array,
    thr: float,
    percentile: bool,
    smooth: bool,
    fwhm: float,
    vmax: float,
    symmetric_colormap: bool,
    disp_mode: str,
    slices: list,
    path2template: str,
    file_name: str,
    save_file: bool
):
    """
    Plot della XAI map sul template MNI152.
    """

    template_file = os.path.join(path2template, "MNI152_T1_2mm.nii.gz")

    if not os.path.exists(template_file):
        raise FileNotFoundError(
            f"Template MNI non trovato: {template_file}"
        )

    mni152_standard = nib.load(template_file)

    if image_roi.shape != mni152_standard.shape:
        raise ValueError(
            f"Shape mismatch: image_roi ha shape {image_roi.shape}, "
            f"ma il template MNI ha shape {mni152_standard.shape}. "
            "La saliency deve essere nella stessa dimensione del template."
        )

    if percentile:
        thr = np.percentile(image_roi, thr)

    image_roi_only_brain = np.zeros(shape=image_roi.shape)

    binary_mask = masking.compute_brain_mask(
        template_file,
        threshold=0.0
    ).get_fdata()

    indices = np.where(binary_mask == 1)
    image_roi_only_brain[indices] = image_roi[binary_mask == 1]

    aff_mni152_standard = mni152_standard.affine

    image_nii = nib.Nifti1Image(
        image_roi_only_brain,
        affine=aff_mni152_standard
    )

    if smooth:
        image_to_plot = image.smooth_img(image_nii, fwhm)
    else:
        image_to_plot = image_nii

    output_file = str(file_name + "_" + disp_mode + ".pdf")

    if save_file:
        plotting.plot_stat_map(
            stat_map_img=image_to_plot,
            bg_img=mni152_standard,
            threshold=thr,
            cmap=plt.cm.viridis,
            vmax=vmax,
            symmetric_cbar=symmetric_colormap,
            display_mode=disp_mode,
            cut_coords=slices,
            resampling_interpolation="continuous",
            draw_cross=False,
            output_file=output_file
        )
    else:
        plotting.plot_stat_map(
            stat_map_img=image_to_plot,
            bg_img=mni152_standard,
            threshold=thr,
            cmap=plt.cm.viridis,
            vmax=vmax,
            symmetric_cbar=symmetric_colormap,
            display_mode=disp_mode,
            cut_coords=slices,
            resampling_interpolation="continuous",
            draw_cross=False,
            output_file=None
        )
        plotting.show()


def plot_mean_saliency_maps(
    mean_saliency_healthy,
    mean_saliency_ad,
    fold,
    modality,
    path2template,
    base_dir="saliency_maps"
):
    """
    Salva i PDF delle saliency medie Healthy e AD
    nelle viste x, y, z.
    """

    output_dir = os.path.join(
        base_dir,
        modality,
        f"fold_{fold}",
        "nilearn_plots"
    )

    os.makedirs(output_dir, exist_ok=True)

    for view in ["x", "y", "z"]:

        if mean_saliency_ad is not None:
            display_brain_maps(
                image_roi=mean_saliency_ad,
                thr=98.0,
                percentile=True,
                smooth=True,
                fwhm=3,
                vmax=None,
                symmetric_colormap=False,
                disp_mode=view,
                slices=list(range(-18,-9,0, 9, 18)),
                path2template=path2template,
                file_name=os.path.join(output_dir, f"{view}_AD"),
                save_file=True
            )

        if mean_saliency_healthy is not None:
            display_brain_maps(
                image_roi=mean_saliency_healthy,
                thr=98.0,
                percentile=True,
                smooth=True,
                fwhm=3,
                vmax=None,
                symmetric_colormap=False,
                disp_mode=view,
                slices=list(range(-18,-9,0, 9, 18)),
                path2template=path2template,
                file_name=os.path.join(output_dir, f"{view}_Healthy"),
                save_file=True
            )

    print(f"PDF XAI salvati in: {output_dir}")


# ============================================================
# IPERPARAMETRI
# ============================================================

epochs = 50
learning_rate = 5e-5
weight_decay = 4e-4
k_folds = 5
batch_size = 12
patience = 18

thresholds = [0.40, 0.45, 0.50, 0.55, 0.60]

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print(f"Device utilizzato: {device}")

# path indicato dal tutor
path2template = "/application/fsl/data/standard/"


# ============================================================
# CROSS VALIDATION
# ============================================================

skf = StratifiedKFold(
    n_splits=k_folds,
    shuffle=True,
    random_state=42
)

splits = list(skf.split(X_input, y))
fold_results = []

all_mean_saliency_healthy = []
all_mean_saliency_ad = []


for fold, (train_ids, val_ids) in enumerate(splits, start=1):
    print(f"\n{'=' * 60}")
    print(f"INIZIO FOLD {fold}/{k_folds} - {ANALISI_CORRENTE}")
    print(f"{'=' * 60}")

    train_ds = CustomDataset(
        X_input[train_ids],
        y[train_ids]
    )

    val_ds = CustomDataset(
        X_input[val_ids],
        y[val_ids]
    )

    train_loader = DataLoader(
        train_ds,
        batch_size=batch_size,
        shuffle=True,
        num_workers=0
    )

    val_loader = DataLoader(
        val_ds,
        batch_size=batch_size,
        shuffle=False,
        num_workers=0
    )

    model = AlzheimerCNN3D(num_classes=2).to(device)

    optimizer = optim.AdamW(
        model.parameters(),
        lr=learning_rate,
        weight_decay=weight_decay
    )

    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=8
    )

    weights_class = torch.tensor([1.0, 1.6]).to(device)

    criterion = nn.CrossEntropyLoss(
        weight=weights_class,
        label_smoothing=0.010
    )

    best_val_f1 = 0.0
    best_epoch = 0
    best_metrics = None
    epochs_no_improve = 0

    train_losses = []
    val_losses = []

    train_accs = []
    val_accs = []

    train_f1s = []
    val_f1s = []

    for epoch in range(epochs):
        train_loss, train_acc, train_bal_acc, train_precision, train_recall, train_f1 = train_epoch(
            model,
            train_loader,
            criterion,
            optimizer,
            device
        )

        best_epoch_threshold = 0.5
        best_epoch_val_f1 = -1.0

        for threshold in thresholds:
            tmp_val_loss, tmp_val_acc, tmp_val_bal_acc, tmp_val_precision, tmp_val_recall, tmp_val_f1 = validate_epoch(
                model,
                val_loader,
                criterion,
                device,
                threshold=threshold
            )

            if tmp_val_f1 > best_epoch_val_f1:
                best_epoch_val_f1 = tmp_val_f1
                best_epoch_threshold = threshold

                val_loss = tmp_val_loss
                val_acc = tmp_val_acc
                val_bal_acc = tmp_val_bal_acc
                val_precision = tmp_val_precision
                val_recall = tmp_val_recall
                val_f1 = tmp_val_f1

        scheduler.step(val_loss)

        train_losses.append(train_loss)
        val_losses.append(val_loss)

        train_accs.append(train_acc)
        val_accs.append(val_acc)

        train_f1s.append(train_f1)
        val_f1s.append(val_f1)

        if val_f1 > best_val_f1:
            best_val_f1 = val_f1
            best_epoch = epoch + 1
            epochs_no_improve = 0

            best_metrics = {
                "fold": fold,
                "best_epoch": best_epoch,
                "best_threshold": best_epoch_threshold,
                "best_val_loss": val_loss,
                "best_val_acc": val_acc,
                "best_val_bal_acc": val_bal_acc,
                "best_val_precision": val_precision,
                "best_val_recall": val_recall,
                "best_val_f1": val_f1
            }

            torch.save(
                model.state_dict(),
                f"best_model_{ANALISI_CORRENTE}_fold{fold}.pth"
            )

        else:
            epochs_no_improve += 1

        if (epoch + 1) % 10 == 0:
            print(
                f"\nFold {fold} | "
                f"Epoca {epoch + 1:03d}/{epochs} | "
                f"T-Loss: {train_loss:.3f} | "
                f"T-Acc: {train_acc:.3f} | "
                f"T-F1: {train_f1:.3f} || "
                f"V-Loss: {val_loss:.3f} | "
                f"V-Acc: {val_acc:.3f} | "
                f"V-F1: {val_f1:.3f} | "
                f"Thr: {best_epoch_threshold:.2f}"
            )

        elif (epoch + 1) % 2 == 0:
            print(".", end="", flush=True)

        if epochs_no_improve >= patience:
            print(
                f"\nEarly stopping attivato "
                f"al fold {fold}, epoca {epoch + 1}"
            )
            break

    fold_results.append(best_metrics)

    print(f"\nFine Fold {fold}")
    print(f"Best Epoch: {best_epoch}")
    print(f"Best Validation F1: {best_val_f1:.3f}")
    print(f"Best Threshold: {best_metrics['best_threshold']}")
    print(f"Validation Acc alla best F1 epoch: {best_metrics['best_val_acc']:.3f}")
    print(f"Validation Balanced Acc alla best F1 epoch: {best_metrics['best_val_bal_acc']:.3f}")

    plot_fold_curves(
        train_losses,
        val_losses,
        train_f1s,
        val_f1s,
        train_accs,
        val_accs,
        fold,
        ANALISI_CORRENTE
    )

    best_model = AlzheimerCNN3D(num_classes=2).to(device)

    best_model.load_state_dict(
        torch.load(
            f"best_model_{ANALISI_CORRENTE}_fold{fold}.pth",
            map_location=device,
            weights_only=True
        )
    )

    best_model.eval()

    mean_saliency_healthy, mean_saliency_ad = compute_mean_saliency_maps_for_fold(
        model=best_model,
        X_data=X_input,
        y=y,
        val_ids=val_ids,
        fold=fold,
        modality=ANALISI_CORRENTE,
        device=device,
        threshold=best_metrics["best_threshold"],
        base_dir="saliency_maps"
    )

    if mean_saliency_healthy is not None:
        all_mean_saliency_healthy.append(mean_saliency_healthy)

    if mean_saliency_ad is not None:
        all_mean_saliency_ad.append(mean_saliency_ad)

    plot_mean_saliency_maps(
        mean_saliency_healthy=mean_saliency_healthy,
        mean_saliency_ad=mean_saliency_ad,
        fold=fold,
        modality=ANALISI_CORRENTE,
        path2template=path2template,
        base_dir="saliency_maps"
    )


# ============================================================
# SALIENCY MEDIA FINALE SU TUTTI I FOLD
# ============================================================

final_xai_dir = os.path.join(
    "saliency_maps",
    ANALISI_CORRENTE,
    "final_mean_maps"
)

os.makedirs(final_xai_dir, exist_ok=True)

final_mean_saliency_healthy = None
final_mean_saliency_ad = None

if len(all_mean_saliency_healthy) > 0:
    final_mean_saliency_healthy = np.mean(
        np.stack(all_mean_saliency_healthy, axis=0),
        axis=0
    )

    np.save(
        os.path.join(final_xai_dir, f"{ANALISI_CORRENTE}_final_mean_saliency_Healthy.npy"),
        final_mean_saliency_healthy
    )

if len(all_mean_saliency_ad) > 0:
    final_mean_saliency_ad = np.mean(
        np.stack(all_mean_saliency_ad, axis=0),
        axis=0
    )

    np.save(
        os.path.join(final_xai_dir, f"{ANALISI_CORRENTE}_final_mean_saliency_AD.npy"),
        final_mean_saliency_ad
    )

print("\nCreazione plot finali XAI su tutti i fold...")

for view in ["x", "y", "z"]:

    if final_mean_saliency_ad is not None:
        display_brain_maps(
            image_roi=final_mean_saliency_ad,
            thr=98.0,
            percentile=True,
            smooth=True,
            fwhm=3,
            vmax=None,
            symmetric_colormap=False,
            disp_mode=view,
            slices=list(range(-18,-9,0, 9, 18)),
            path2template=path2template,
            file_name=os.path.join(final_xai_dir, f"{view}_AD_final"),
            save_file=True
        )

    if final_mean_saliency_healthy is not None:
        display_brain_maps(
            image_roi=final_mean_saliency_healthy,
            thr=98.0,
            percentile=True,
            smooth=True,
            fwhm=3,
            vmax=None,
            symmetric_colormap=False,
            disp_mode=view,
            slices=list(range(-18,-9,0, 9, 18)),
            path2template=path2template,
            file_name=os.path.join(final_xai_dir, f"{view}_Healthy_final"),
            save_file=True
        )

print(f"Plot finali salvati in: {final_xai_dir}")


# ============================================================
# RISULTATI FINALI
# ============================================================

print("\n" + "=" * 60)
print(f"RISULTATI FINALI CROSS-VALIDATION - {ANALISI_CORRENTE}")
print("=" * 60)

best_f1s = [r["best_val_f1"] for r in fold_results]
best_accs = [r["best_val_acc"] for r in fold_results]
best_bal_accs = [r["best_val_bal_acc"] for r in fold_results]
best_precs = [r["best_val_precision"] for r in fold_results]
best_recs = [r["best_val_recall"] for r in fold_results]
best_thrs = [r["best_threshold"] for r in fold_results]

print(f"Best F1 per fold: {[round(x, 3) for x in best_f1s]}")
print(f"ACC alla best F1 epoch per fold: {[round(x, 3) for x in best_accs]}")
print(f"Balanced ACC alla best F1 epoch per fold: {[round(x, 3) for x in best_bal_accs]}")
print(f"PREC alla best F1 epoch per fold: {[round(x, 3) for x in best_precs]}")
print(f"REC alla best F1 epoch per fold: {[round(x, 3) for x in best_recs]}")
print(f"Threshold migliori per fold: {best_thrs}")

print(f"\nF1 medio: {np.mean(best_f1s):.3f} ± {np.std(best_f1s):.3f}")
print(f"ACC media: {np.mean(best_accs):.3f} ± {np.std(best_accs):.3f}")
print(f"Balanced ACC media: {np.mean(best_bal_accs):.3f} ± {np.std(best_bal_accs):.3f}")
print(f"PREC media: {np.mean(best_precs):.3f} ± {np.std(best_precs):.3f}")
print(f"REC media: {np.mean(best_recs):.3f} ± {np.std(best_recs):.3f}")


# ============================================================
# REPORT FINALE AGGREGATO
# ============================================================

def generate_final_text_report(
    X_data,
    y,
    device,
    modality,
    splits,
    fold_results
):
    all_preds = []
    all_labels = []

    for fold, (_, val_ids) in enumerate(splits, start=1):
        model = AlzheimerCNN3D(num_classes=2).to(device)

        model.load_state_dict(
            torch.load(
                f"best_model_{modality}_fold{fold}.pth",
                map_location=device,
                weights_only=True
            )
        )

        model.eval()

        threshold = fold_results[fold - 1]["best_threshold"]

        val_ds = CustomDataset(
            X_data[val_ids],
            y[val_ids]
        )

        val_loader = DataLoader(
            val_ds,
            batch_size=1,
            shuffle=False
        )

        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs = inputs.to(device)

                outputs = model(inputs)

                predicted = predict_with_threshold(
                    outputs,
                    threshold=threshold
                )

                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(labels.numpy())

    cm = confusion_matrix(all_labels, all_preds)

    print(f"\n{'=' * 20} REPORT FINALE {modality} {'=' * 20}")

    print(
        classification_report(
            all_labels,
            all_preds,
            target_names=["Healthy", "AD"],
            zero_division=0
        )
    )

    print(f"Accuracy finale aggregata: {accuracy_score(all_labels, all_preds):.3f}")
    print(f"Balanced accuracy finale aggregata: {balanced_accuracy_score(all_labels, all_preds):.3f}")
    print(f"Macro F1 finale aggregato: {f1_score(all_labels, all_preds, average='macro', zero_division=0):.3f}")

    plt.figure(figsize=(6, 5))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Healthy", "AD"],
        yticklabels=["Healthy", "AD"]
    )

    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(f"Final Confusion Matrix: {modality}")
    plt.tight_layout()
    plt.show()


generate_final_text_report(
    X_input,
    y,
    device,
    ANALISI_CORRENTE,
    splits,
    fold_results
)
```
