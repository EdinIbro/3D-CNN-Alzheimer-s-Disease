```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

import numpy as np
import torch.nn.functional as F
import pickle
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)


# ==========================================
# 1. CARICAMENTO E FILTRAGGIO DATASET
# ==========================================
with open("Dataset_Edin.pkl", "rb") as f:
    data_dict = pickle.load(f)

X_fa_raw = np.array(data_dict["FA"])
X_md_raw = np.array(data_dict["MD"])
y_raw = np.array(data_dict["Labels"])

mask = (y_raw == 0) | (y_raw == 2)

X_fa = X_fa_raw[mask]
X_md = X_md_raw[mask]
y = y_raw[mask]

y[y == 2] = 1

ANALISI_CORRENTE = "MD"  # cambia in "MD" per analizzare MD
X_input = X_md if ANALISI_CORRENTE == "MD" else X_fa

print(f"Analisi corrente: {ANALISI_CORRENTE}")
print(f"Dataset: {len(y)} campioni")
print(f"Sani: {np.sum(y == 0)} | Alzheimer: {np.sum(y == 1)}")


# ==========================================
# 2. DATASET SENZA RESIZE
# ==========================================
class AlzheimerDataset(Dataset):
    def __init__(self, data, labels):
        self.data = data
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def normalize(self, data):
        data = np.nan_to_num(data)

        dmin = data.min()
        dmax = data.max()

        if dmax > dmin:
            return (data - dmin) / (dmax - dmin)

        return data

    def __getitem__(self, idx):
        vol = self.data[idx]
        vol_norm = self.normalize(vol)

        x = torch.from_numpy(vol_norm).float().unsqueeze(0)
        y_label = torch.tensor(self.labels[idx], dtype=torch.long)

        return x, y_label


def collate_fn_no_resize(batch):
    imgs, labels = zip(*batch)

    max_d = max(img.shape[1] for img in imgs)
    max_h = max(img.shape[2] for img in imgs)
    max_w = max(img.shape[3] for img in imgs)

    padded_imgs = []

    for img in imgs:
        d_pad = max_d - img.shape[1]
        h_pad = max_h - img.shape[2]
        w_pad = max_w - img.shape[3]

        img = F.pad(img, (0, w_pad, 0, h_pad, 0, d_pad))
        padded_imgs.append(img)

    return torch.stack(padded_imgs), torch.tensor(labels, dtype=torch.long)


# ==========================================
# 3. MODELLO CNN 3D
# ==========================================
class AlzheimerCNN3D(nn.Module):
    def __init__(self, num_classes=2):
        super(AlzheimerCNN3D, self).__init__()

        self.features = nn.Sequential(
            nn.Conv3d(1, 32, kernel_size=3, padding=1),
            nn.BatchNorm3d(32),
            nn.ReLU(),
            nn.MaxPool3d(2),

            nn.Conv3d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm3d(64),
            nn.ReLU(),
            nn.MaxPool3d(2),

            nn.Conv3d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm3d(128),
            nn.ReLU(),
            nn.MaxPool3d(2),

            nn.Conv3d(128, 256, kernel_size=3, padding=1),
            nn.BatchNorm3d(256),
            nn.ReLU()
        )

        self.gap = nn.AdaptiveAvgPool3d(1)

        self.classifier = nn.Sequential(
            nn.Dropout(0.6),
            nn.Linear(256, 256),
            nn.ReLU(),
            nn.LayerNorm(256),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.gap(x)
        x = torch.flatten(x, 1)
        x = self.classifier(x)
        return x


# ==========================================
# 4. METRICHE CON SKLEARN
# ==========================================
def compute_metrics(y_true, y_pred):
    acc = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    return acc, precision, recall, f1


# ==========================================
# 5. TRAINING E VALIDATION
# ==========================================
def train_epoch(model, loader, criterion, optimizer, device):
    model.train()

    running_loss = 0.0
    all_preds = []
    all_labels = []

    for inputs, labels in loader:
        inputs = inputs.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(inputs)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        _, predicted = outputs.max(1)

        running_loss += loss.item()

        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

    avg_loss = running_loss / len(loader)
    acc, precision, recall, f1 = compute_metrics(all_labels, all_preds)

    return avg_loss, acc, precision, recall, f1


def validate_epoch(model, loader, criterion, device):
    model.eval()

    running_loss = 0.0
    all_preds = []
    all_labels = []

    with torch.no_grad():
        for inputs, labels in loader:
            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model(inputs)
            loss = criterion(outputs, labels)

            _, predicted = outputs.max(1)

            running_loss += loss.item()

            all_preds.extend(predicted.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    avg_loss = running_loss / len(loader)
    acc, precision, recall, f1 = compute_metrics(all_labels, all_preds)

    return avg_loss, acc, precision, recall, f1


# ==========================================
# 6. PLOT PER OGNI FOLD
# ==========================================
def plot_fold_curves(train_losses, val_losses, train_accs, val_accs, fold, modality):
    epochs_range = range(1, len(train_losses) + 1)

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    axes[0].plot(epochs_range, train_losses, label="Training Loss")
    axes[0].plot(epochs_range, val_losses, label="Validation Loss")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss")
    axes[0].set_title(f"Fold {fold} - Loss")
    axes[0].legend()
    axes[0].grid(True)

    axes[1].plot(epochs_range, train_accs, label="Training Accuracy")
    axes[1].plot(epochs_range, val_accs, label="Validation Accuracy")
    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Accuracy")
    axes[1].set_title(f"Fold {fold} - Accuracy")
    axes[1].legend()
    axes[1].grid(True)

    plt.suptitle(f"{modality} - Training vs Validation - Fold {fold}")
    plt.tight_layout()
    plt.show()


# ==========================================
# 7. CROSS-VALIDATION
# ==========================================
epochs = 100
learning_rate = 1e-4
k_folds = 5
batch_size = 16

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

skf = StratifiedKFold(
    n_splits=k_folds,
    shuffle=True,
    random_state=42
)

fold_results = []

for fold, (train_ids, val_ids) in enumerate(skf.split(X_input, y), start=1):
    print(f"\n{'=' * 60}")
    print(f"INIZIO FOLD {fold}/{k_folds} - {ANALISI_CORRENTE}")
    print(f"{'=' * 60}")

    train_ds = AlzheimerDataset(
        X_input[train_ids],
        y[train_ids]
    )

    val_ds = AlzheimerDataset(
        X_input[val_ids],
        y[val_ids]
    )

    train_loader = DataLoader(
        train_ds,
        batch_size=batch_size,
        shuffle=True,
        collate_fn=collate_fn_no_resize
    )

    val_loader = DataLoader(
        val_ds,
        batch_size=batch_size,
        shuffle=False,
        collate_fn=collate_fn_no_resize
    )

    model = AlzheimerCNN3D(num_classes=2).to(device)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=learning_rate,
        weight_decay=1e-4
    )

    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.2,
        patience=10
    )

    class_weights = torch.tensor([1.0, 1.8]).to(device)

    criterion = nn.CrossEntropyLoss(
        weight=class_weights,
    )

    best_val_acc = 0.0
    best_epoch = 0

    train_losses = []
    val_losses = []
    train_accs = []
    val_accs = []

    train_precisions = []
    val_precisions = []
    train_recalls = []
    val_recalls = []
    train_f1s = []
    val_f1s = []

    for epoch in range(epochs):
        train_loss, train_acc, train_precision, train_recall, train_f1 = train_epoch(
            model,
            train_loader,
            criterion,
            optimizer,
            device
        )

        val_loss, val_acc, val_precision, val_recall, val_f1 = validate_epoch(
            model,
            val_loader,
            criterion,
            device
        )

        scheduler.step(val_loss)

        train_losses.append(train_loss)
        val_losses.append(val_loss)

        train_accs.append(train_acc)
        val_accs.append(val_acc)

        train_precisions.append(train_precision)
        val_precisions.append(val_precision)

        train_recalls.append(train_recall)
        val_recalls.append(val_recall)

        train_f1s.append(train_f1)
        val_f1s.append(val_f1)

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_epoch = epoch + 1

            torch.save(
                model.state_dict(),
                f"best_model_{ANALISI_CORRENTE}_fold{fold}.pth"
            )

        if (epoch + 1) % 10 == 0:
            print(
                f"\nFold {fold} | Epoca {epoch + 1:03d}/{epochs} | "
                f"T-Loss: {train_loss:.3f} | "
                f"T-Acc: {train_acc:.3f} | "
                f"T-Prec: {train_precision:.3f} | "
                f"T-Rec: {train_recall:.3f} | "
                f"T-F1: {train_f1:.3f} || "
                f"V-Loss: {val_loss:.3f} | "
                f"V-Acc: {val_acc:.3f} | "
                f"V-Prec: {val_precision:.3f} | "
                f"V-Rec: {val_recall:.3f} | "
                f"V-F1: {val_f1:.3f}"
            )
        elif (epoch + 1) % 2 == 0:
            print(".", end="", flush=True)

    fold_results.append({
        "fold": fold,
        "best_epoch": best_epoch,
        "best_val_f1": max(val_f1),
        "best_val_acc": max(val_accs),
        "best_val_precision": max(val_precisions),
        "best_val_recall": max(val_recalls)
    })

    print(f"\nFine Fold {fold}")
    print(f"Best Epoch: {best_epoch}")
    print(f"Best Validation Acc: {best_val_acc:.3f}")

    plot_fold_curves(
        train_losses,
        val_losses,
        train_accs,
        val_accs,
        fold,
        ANALISI_CORRENTE
    )


# ==========================================
# 8. RISULTATI FINALI CROSS-VALIDATION
# ==========================================
print("\n" + "=" * 60)
print(f"RISULTATI FINALI CROSS-VALIDATION - {ANALISI_CORRENTE}")
print("=" * 60)

best_f1s = [r["best_val_f1"] for r in fold_results]
best_accs = [r["best_val_acc"] for r in fold_results]
best_precs = [r["best_val_precision"] for r in fold_results]
best_recs = [r["best_val_recall"] for r in fold_results]

print(f"Best F1 per fold: {[round(x, 3) for x in best_f1s]}")
print(f"Best ACC per fold: {[round(x, 3) for x in best_accs]}")
print(f"Best PREC per fold: {[round(x, 3) for x in best_precs]}")
print(f"Best REC per fold: {[round(x, 3) for x in best_recs]}")

print(f"\nF1 medio: {np.mean(best_f1s):.3f} ± {np.std(best_f1s):.3f}")
print(f"ACC media: {np.mean(best_accs):.3f} ± {np.std(best_accs):.3f}")
print(f"PREC media: {np.mean(best_precs):.3f} ± {np.std(best_precs):.3f}")
print(f"REC media: {np.mean(best_recs):.3f} ± {np.std(best_recs):.3f}")


# ==========================================
# 9. REPORT FINALE AGGREGATO
# ==========================================
def generate_final_text_report(X_data, y, device, modality):
    all_preds = []
    all_labels = []

    skf = StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    )

    for fold, (_, val_ids) in enumerate(skf.split(X_data, y), start=1):
        model = AlzheimerCNN3D(num_classes=2).to(device)

        model.load_state_dict(
            torch.load(
                f"best_model_{modality}_fold{fold}.pth",
                map_location=device
            )
        )

        model.eval()

        val_ds = AlzheimerDataset(
            X_data[val_ids],
            y[val_ids]
        )

        val_loader = DataLoader(
            val_ds,
            batch_size=1,
            shuffle=False,
            collate_fn=collate_fn_no_resize
        )

        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs = inputs.to(device)

                outputs = model(inputs)
                _, predicted = outputs.max(1)

                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(labels.numpy())

    cm = confusion_matrix(all_labels, all_preds)

    print(f"\n{'=' * 20} REPORT FINALE {modality} {'=' * 20}")
    print(
        classification_report(
            all_labels,
            all_preds,
            target_names=["Sano", "AD"],
            zero_division=0
        )
    )

    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Sano", "AD"],
        yticklabels=["Sano", "AD"]
    )
    plt.xlabel("Predetto")
    plt.ylabel("Reale")
    plt.title(f"Matrice di Confusione Finale: {modality}")
    plt.show()


generate_final_text_report(
    X_input,
    y,
    device,
    ANALISI_CORRENTE
) 
```
